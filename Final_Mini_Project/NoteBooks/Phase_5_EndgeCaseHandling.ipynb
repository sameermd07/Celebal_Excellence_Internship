# phase 5 : edge case handling

validate common edge cases in the order data.

```python
import pandas as pd
```

```python
customers_df = pd.read_csv("/Volumes/mini_project/default/cleaned_data/customers_clean.csv")
products_df = pd.read_csv("/Volumes/mini_project/default/cleaned_data/products_clean.csv")
orders_df = pd.read_csv("/Volumes/mini_project/default/cleaned_data/orders_clean.csv")
order_items_df = pd.read_csv("/Volumes/mini_project/default/cleaned_data/order_items_clean.csv")
```

## 1. check order_items with invalid order_id

find order items that reference an order_id which does not exist in the orders table.

```python
def test_invalid_order_id(orders_df, order_items_df):

    invalid_orders = order_items_df[
        ~order_items_df["order_id"].isin(orders_df["order_id"])
    ]

    return invalid_orders
```

```python
test_invalid_order_id(orders_df, order_items_df)
```

## 2. check discount greater than 100

```python
def test_invalid_discount(order_items_df):

    invalid_discount = order_items_df[
        order_items_df["discount_percent"] > 100
    ]

    return invalid_discount
```

```python
test_invalid_discount(order_items_df)
```

## 3. check quantity equal to zero

```python
def test_zero_quantity(order_items_df):

    zero_quantity = order_items_df[
        order_items_df["quantity"] == 0
    ]

    return zero_quantity
```

```python
test_zero_quantity(order_items_df)
```

## 4. check future order dates

```python
from datetime import datetime

def test_future_orders(orders_df):
    orders_df["order_date"] = pd.to_datetime(orders_df["order_date"])
    
    future_orders = orders_df[
        orders_df["order_date"] > pd.Timestamp(datetime.now())
    ]
    return future_orders
```

```python
test_future_orders(orders_df)
```
