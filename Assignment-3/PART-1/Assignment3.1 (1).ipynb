### **Finding Duplicate Orders**
- Check data quality.
- GROUP BY, HAVING

```sql
select `Order ID`,`Product ID`,count(*) as cnt from sample_superstore group by `Order ID`,`Product ID` having count(*)>1;
```

**Why I Grouped by Both IDs**
- One Order, Multiple Items: A customer can buy a chair and a pen. This creates two rows with the same Order ID but different Product IDs. This is normal.

- The Error: If the exact same Product ID appears twice under the same Order ID, the system accidentally double-counted the item.

- The Solution: Grouping by both IDs skips normal orders and only flags true data glitches.

### **Loss-Making `Orders`**

```sql
select `Order ID`,`Product ID`,round(sum(profit),2) as total_loss from sample_superstore group by `Order ID`,`Product ID` having sum(profit)<0 order by total_loss asc limit 10;
```

- groups by both order_id and product_id to look at every single item line.
- calculates the total profit for that line and filters for anything less than 0 (having sum(profit)<0).
- sorts it so your biggest financial losses appear right at the top.
