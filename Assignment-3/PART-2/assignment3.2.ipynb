Q5>Rank all customers based on total sales. (Window Function)  

Here also we can use the same q3's cte any how it gave us the total sales of each customer and we can apply window functions such as `dense_rank()`,`rank()` to give them ranking based on their `total salea`

```sql
with res as (select Customer_ID, sum(Sales) as total_sales from orders group by Customer_ID)
select *,dense_rank() over(order by total_sales desc) as rank from res limit 10;
```

```sql
with res as (select Customer_ID, sum(Sales) as total_sales from orders group by Customer_ID)
select *,rank() over(order by total_sales desc) as rank from res limit 10;
```

- `RANK() `skips ranks after a tie.

- `DENSE_RANK() `does NOT skip ranks after a tie; the next rank is always the very next  number.

Q6>Assign row numbers to each order within a customer. (Window Function + PARTITION BY)

Q7>Display top 3 customers based on total sales. (Window Function)  

### we have three tasks to perform here 
- 1>finding total sales of all the customers we can use the same code as the `Q4`
- 2>rankig them based on their `total sales`
- 3>filter the rows with rank<=3

```sql
with res1 as (select Customer_ID, round(sum(Sales),2) as total_sales from orders group by Customer_ID),
res2 as (select *,dense_rank() over(order by total_sales desc) as rank from res1)

select * from res2 where rank <= 3;
```

# Mini Project: Customer Sales Insights 

1>Who are the top 5 customers?  

```sql
with res as ( select customer_id,round(sum(sales), 2) as total_sales from orders group by customer_id
) ,
res2 as (select distinct b.customer_name,a.total_sales,
dense_rank() over(order by a.total_sales desc) as sales_rank
from res a 
join customers b on a.customer_id = b.customer_id)
select * from res2 where sales_rank<=5;
```

- Purpose: Identify the top 5 customers based on total sales.
- SQL Concept Used: CTE + JOIN + Window Function (DENSE_RANK).
- Why Suitable: Ranking helps find the highest revenue-generating customers.

2>Who are the bottom 5 customers?  

- same as above code but when applying the window function just use `order by total_sales asc instead desc`

```sql
with res as ( select customer_id,round(sum(sales), 2) as total_sales from orders group by customer_id
) ,
res2 as (select distinct b.customer_name,a.total_sales,
dense_rank() over(order by a.total_sales asc) as sales_rank
from res a 
join customers b on a.customer_id = b.customer_id)
select * from res2 where sales_rank<=5;
```

- Purpose: Identify the bottom 5 customers based on total sales.
- SQL Concept Used: CTE + JOIN + Window Function (DENSE_RANK).
- Why Suitable: Ranking in ascending order highlights the least contributing customers.

3>Which customers made only one order?  

```sql
select b.customer_name,count(distinct o.order_id) as total_orders
from orders o
join (select customer_id, max(customer_name) as customer_name from customers 
group by customer_id
) b on o.customer_id = b.customer_id
group by o.customer_id, b.customer_name
having count(distinct o.order_id) = 1 limit 10;
```

- Purpose: Find customers who placed exactly one order.
- SQL Concept Used: JOIN + GROUP BY + HAVING.
- Why Suitable: HAVING filters customers based on aggregated order counts.

4>Which customers have above-average sales?

```sql
with customer_sales as (
select customer_id, round(sum(sales),2) as total_sales from orders group by customer_id
)
select c.customer_name, cs.total_sales
from customer_sales cs
join (select customer_id, max(customer_name) as customer_name from customers group by customer_id) c on cs.customer_id = c.customer_id
where cs.total_sales > (select avg(total_sales) from customer_sales) limit 10;
```

- Purpose: Find customers whose total sales exceed the average customer sales.
- SQL Concept Used: CTE + JOIN + Subquery.
- Why Suitable: Subquery calculates average sales for comparison.

5>What is the highest order value per customer? 

```sql
select c.customer_name, max(o.sales) as highest_order_value
from orders o
join (select customer_id, max(customer_name) as customer_name from customers group by customer_id) c on o.customer_id = c.customer_id
group by c.customer_name limit 10;
```

- Purpose: Determine the maximum order amount placed by each customer.
- SQL Concept Used: JOIN + GROUP BY + Aggregate Function (MAX).
- Why Suitable: MAX() efficiently retrieves the highest sales value for each customer.
