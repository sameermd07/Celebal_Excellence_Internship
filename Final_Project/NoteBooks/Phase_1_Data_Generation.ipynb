```python
from faker import Faker
from datetime import datetime
import pandas as pd
import random
```

```python
fake=Faker()
random.seed(42)
Faker.seed(42)
```

```python
NUM_CUSTOMERS = 800
NUM_PRODUCTS = 700
NUM_ORDERS = 900
NUM_ORDER_ITEMS = 2000

CUSTOMER_TYPES = [
    "REGULAR",
    "PREMIUM",
    "VIP"
]

CATEGORIES = [
    "Electronics",
    "Clothing",
    "Home",
    "Books"
]

ORDER_STATUS = [
    "PLACED",
    "SHIPPED",
    "DELIVERED",
    "CANCELLED",
    "RETURNED"
]
```

```python
def generate_customers(num_customers):
    customers = []
    cities = [
        "Hyderabad",
        "Bangalore",
        "Chennai",
        "Mumbai",
        "Delhi",
        "Pune",
        "Kolkata",
        "Ahmedabad"
    ]
    for customer_id in range(1, num_customers + 1):
        customer = {}
        customer["customer_id"] = customer_id
        customer["customer_name"] = fake.name()
        email = fake.email()
        # Creating invaid emails around 2%
        if random.random() < 0.02:
            if random.choice([True, False]):
                email = email.replace("@", "")
            else:
                email = email.split("@")[0]

        customer["email"] = email
        customer["city"] = random.choice(cities)
        customer["customer_type"] = random.choice(CUSTOMER_TYPES)
        customers.append(customer)
    return pd.DataFrame(customers)
```

```python

PRODUCTS = {
    "Electronics": [
        "iPhone 15",
        "Samsung Galaxy S24",
        "Dell Laptop",
        "HP Laptop",
        "Sony Headphones",
        "Bluetooth Speaker",
        "Smart Watch",
        "Wireless Mouse",
        "Mechanical Keyboard",
        "Power Bank",
        "USB Cable",
        "Monitor",
        "Webcam",
        "Gaming Mouse",
        "Printer"
    ],

    "Clothing": [
        "T-Shirt",
        "Jeans",
        "Hoodie",
        "Jacket",
        "Sneakers",
        "Formal Shirt",
        "Kurta",
        "Track Pants",
        "Cap",
        "Socks",
        "Sweatshirt",
        "Polo Shirt"
    ],

    "Home": [
        "Dining Table",
        "Sofa",
        "Wall Clock",
        "Microwave",
        "Mixer Grinder",
        "Vacuum Cleaner",
        "Water Bottle",
        "Chair",
        "Study Table",
        "Curtains",
        "LED Lamp",
        "Pressure Cooker"
    ],

    "Books": [
        "Python Programming",
        "SQL Cookbook",
        "Machine Learning Basics",
        "Data Engineering Guide",
        "Clean Code",
        "Atomic Habits",
        "Deep Learning",
        "Operating Systems",
        "Computer Networks",
        "DSA Handbook"
    ]
}
```

```python
def generate_products(num_products):
    products = []
    for product_id in range(1, num_products + 1):
        #coosing a random category
        category = random.choice(CATEGORIES)
        product_name = random.choice(PRODUCTS[category])
        probability = random.random()
        # 5% of dirty data with extra spaces
        if probability < 0.05:
            product_name = "   " + product_name + "   "
        # 5% of dirty data with upper/lower case
        elif probability < 0.10:
            product_name = product_name.swapcase()
        #generating random nums between 100-5000
        price = round(random.uniform(100, 50000), 2)
        products.append({
            "product_id": product_id,
            "product_name": product_name,
            "category": category,
            "unit_price": price
        })

    return pd.DataFrame(products)
```

```python
def generate_orders(num_orders, num_customers):
    orders = []
    for order_id in range(1, num_orders + 1):
        #5% null customer id
        if random.random() < 0.05:
            customer_id = None
        else:
            customer_id = random.randint(1, num_customers)
        #date
        order_date = fake.date_time_between(
            start_date="-2y",
            end_date="now"
        )
        # 5% Wrong Date Format
        if random.random() < 0.05:
            order_date = order_date.strftime("%d-%m-%Y %H:%M:%S")
        else:
            order_date = order_date.strftime("%Y-%m-%d %H:%M:%S")
        status = random.choice(ORDER_STATUS)
        orders.append({
            "order_id": order_id,
            "customer_id": customer_id,
            "order_date": order_date,
            "status": status
        })

    return pd.DataFrame(orders)
```

```python
def generate_order_items(orders_df, num_products):
    order_items = []
    order_item_id = 1
    for order_id in orders_df["order_id"]:
        # Every order has 1 to 5 products
        number_of_products = random.randint(1, 5)
        # Prevent duplicate product in same order
        selected_products = random.sample(
            range(1, num_products + 1),
            number_of_products
        )
        for product_id in selected_products:
            quantity = random.randint(1, 5)
            # 3% Negative Quantity
            if random.random() < 0.03:
                quantity *= -1
            discount = round(random.uniform(0, 100), 2)
            order_items.append({
                "order_item_id": order_item_id,
                "order_id": order_id,
                "product_id": product_id,
                "quantity": quantity,
                "discount_percent": discount
            })
            order_item_id += 1
    return pd.DataFrame(order_items)
```
