Q1. What are the key limitations of traditional MapReduce that make Spark a preferred choice for modern big data processing?

Traditional MapReduce has some problems:

- It is slow because it stores data on the disk after every step.
- It cannot process data in memory, so it takes more time.
- It needs more coding, so it is harder to use.

Why Spark is better:

- Spark is much faster because it uses memory (RAM).
- It is easy to write programs.
- It can handle batch processing, real-time data.
- It saves time and gives results quickly.

Q11: Explain the "Shuffle" process that occurs during a grouping operation. Why is it considered a wide transformation? 

A shuffle occurs when Spark redistributes data across different partitions during operations such as:
- groupBy()
- join()
- distinct()
- reduceByKey()
- Since data must move between executors over the network, shuffle is considered an expensive operation.
- It is called a wide transformation because records from multiple partitions are combined to produce the final result.

Q14: In the context of cleaning a dataset, what is the risk of using inferSchema=true when your source data contains messy or inconsistent date formats? 


When a dataset contains multiple date formats, Spark may incorrectly infer the data type.

Possible issues include:

- Incorrect schema detection.
- Date parsing failures.
- Null values in date columns.
- Inconsistent processing results.

For production environments, it is generally recommended to define the schema explicitly instead of relying only on `inferSchema=true`.
